```python
def getAccountSolBalance(account_address):
    """ return balance in imports of the account address 
    ! does not seem to require API credentials... so its just free and unlimited ?
    """
    url = "https://nd-326-444-187.p2pify.com/9de47db917d4f69168e3fed02217d15b/"

    payload = {
        "id": 1,
        "jsonrpc": "2.0",
        "method": "getAccountInfo",
        "params": [

                account_address,
            {
                "encoding": "jsonParsed",
                "commitment": "finalized"
            }
        ]
    }
    headers = {
        "accept": "application/json",
        "content-type": "application/json"
    }

    response = requests.post(url, json=payload, headers=headers)

    return float(response.json()['result']['value']['lamports'])/10**9

response = getAccountSolBalance(pubkey_perso)

response
```
